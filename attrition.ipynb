### target analysis

- What is the distribution of employee attrition?

```python
fig = px.histogram(df,x="Attrition",histnorm="percent",text_auto=".1f",title="Employee Attrition Distribution (%)")

fig.update_traces(texttemplate="%{y:.1f}%")
fig
```

- Approximately 80% of employees stayed with the company, while around 20% left. This indicates that employee attrition exists but affects a minority of the workforce.

### Univariate Analysis

```python
df.select_dtypes(include='number').columns
```

- Numerical Features

```python
for col in df.select_dtypes(include='number').columns.drop('Employee_ID'):
    fig = px.box(
        df,
        y=col,
        title=f"{col} Distribution"
    )
    fig.show()
```

- Categorical Features 

```python
for col in df.select_dtypes(include="object").columns:
    fig = px.histogram(
        df,
        x=col,
        text_auto=True,
        title=f"{col} Distribution",
        category_orders={
            col: df[col].value_counts().index.tolist()
        }
    )

    fig.show()
```

### Bivariate Anakysis

- Attrition Vs Department

```python
dep = df.groupby("Department")["Attrition"].value_counts(normalize=True).reset_index(name="Percentage")

fig = px.bar(dep, x="Department", y="Percentage", color="Attrition", text_auto=".1%")
fig.show()
```

- Attrition rates are very similar across all departments, suggesting that department is not a strong factor associated with employee attrition.

### Attrition Vs Overtime

```python
ot = df.groupby("Overtime")["Attrition"].value_counts(normalize=True).reset_index(name="Percentage")

fig = px.bar(
    ot,
    x="Overtime",
    y="Percentage",
    color="Attrition",
    text_auto=".1%"
)

fig.show()
```

- Overtime shows a noticeable association with attrition, with employees working overtime having a higher proportion of attrition.

### Attrition Vs Monthly income

```python
fig = px.box(
    df,
    x="Attrition",
    y="Monthly_Income",
    title="Monthly Income vs Attrition"
)

fig.show()
```

### Attrition Vs Job Satisfaction 

```python
fig = px.box(
    df,
    x="Attrition",
    y="Job_Satisfaction",
    title="Job Satisfaction vs Attrition"
)

fig.show()
```

### Job Role vs Attrition

```python
order = df[df["Attrition"] == "Yes"]["Job_Role"].value_counts().index.tolist() # for ordering the job roles based on the number of attrition cases

fig = px.histogram(
    df,
    x="Job_Role",
    color="Attrition",
    text_auto=True,
    category_orders={"Job_Role": order}
)

fig.show()
```

- Attrition rates are relatively similar across job roles, with Assistant having the highest attrition rate and Manager the lowest. However, the differences are not very large.

### Performance Rating vs Attrition
